```python
short_count = (df['text'].str.strip().str.len() < 23).sum()
pct_short = 100 * short_count / len(df)
print(f"Textes < 23 caracteres : {short_count} ({pct_short:.1f}% du dataset)")
print()

textes_courts = df[df['text'].str.strip().str.len() < 23]['text']
print(textes_courts.to_string())
```

```python
short_count = (df['text'].str.strip().str.len() < 5).sum()
pct_short = 100 * short_count / len(df)
print(f"Textes < 5 caracteres : {short_count} ({pct_short:.1f}% du dataset)")
print()

textes_courts = df[df['text'].str.strip().str.len() < 5]['text']
print(textes_courts.to_string())
```

```python
short = df[df['text'].str.strip().str.split().str.len() == 1]

print(f"Lignes avec un seul mot : {len(short)}")
print()
print(short['text'].to_string())
```

```python
guillemets = df[df['text'].str.contains('"', na=False)]

print(f"Lignes avec guillemets : {len(guillemets)}")
print()
print(guillemets['text'].to_string())
```

```python
# lignes avec des balises HTML
html = df[df['text'].str.contains(r'<[^>]+>', regex=True, na=False)]
print(f"Lignes avec balises HTML : {len(html)}")
print()
print(html['text'].to_string())

print()
print("-" * 50)

# lignes avec des URLs
urls = df[df['text'].str.contains(r'http[s]?://\S+', regex=True, na=False)]
print(f"Lignes avec URLs : {len(urls)}")
print()
print(urls['text'].to_string())
```

# Cleaning

```python
df = df.drop(columns=['sources', 'len'])
df.head(10)
```

```python
# Supprimer les NaN
df = df.dropna(subset=['text'])

# Supprimer les chaines vides ou whitespace
df = df[df['text'].str.strip() != '']

print()
print(f"Lignes apres : {len(df)}")
```

```python
# lignes qui contiennent uniquement des chiffres
only_numbers = df[df['text'].str.strip().str.match(r'^\d+$')]

print(f"Lignes avec uniquement des nombres : {len(only_numbers)}")
print()
print(only_numbers['text'].value_counts().to_string())
```

```python
#supprimer les nombres
df = df[~df['text'].str.strip().str.match(r'^\d+$')]

print(f"Lignes apres : {len(df)}")
df.head(10)
```

```python
#supprimer les lignes avec un seul mot
df = df[df['text'].str.strip().str.split().str.len() > 1]

print(f"Lignes apres : {len(df)}")
df.head(10)
```

```python
#on supprime les guillemets du texte

df['text'] = df['text'].str.replace('"', '', regex=False)

df.head(10)
```

```python
#suppression des balises HTML
df = df[~df['text'].str.contains(r'<[^>]+>', regex=True)]

print(f"Lignes apres : {len(df)}")
df.head(10)
```
